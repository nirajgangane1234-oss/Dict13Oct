```python
### WAP to get unique element list from above list without changing data's position

l = [5, 4, 5, 4, 6, 7, 8, 4, 3, 5, 3, 4]

unique_list = []

for i in l:
    if i not in unique_list:
        unique_list.append(i)

print(unique_list)
```

```python
### Find freq of every elementd from above input list dict

l = [5, 4, 5, 4, 6, 7, 8, 4, 3, 5, 3, 4]

freq_ele = {}

for i in l:
    if i in freq_ele:
        freq_ele[i] += 1
    else:
        freq_ele[i] = 1

print(freq_ele)
```

```python
### freq of each and every character from the dict

s = "I love python programming"

freq_char = {}

for i in s:
    if i in freq_char:
        freq_char[i] += 1
    else:
        freq_char[i] = 1

print(freq_char)
```
